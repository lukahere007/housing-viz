Walks through every plot in `housing_viz.py`. The data is the Kaggle dataset
[*Global Housing Market Analysis 2015-2024*](https://www.kaggle.com/datasets/atharvasoundankar/global-housing-market-analysis-2015-2024)
(20 countries x 10 years). Put the CSV in `data/`, or set up a Kaggle API token so `kagglehub` can download it.

**Read the plots as a visualisation exercise, not as facts about housing markets.** The values behave like
independent random draws (see the last section and the README), so country rankings and trends here carry no
real-world meaning.

*Outputs are not saved in this notebook: the data needs a Kaggle download, which was unavailable when the repo
was last tidied. The PNGs in `figs/` come from the author's original run.*

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # make housing_viz importable from notebooks/

import matplotlib.pyplot as plt
import seaborn as sns
import housing_viz as hv

df = hv.load_housing_data()
df.head()
```

## 1. House-price index, all countries

```python
plt.figure(figsize=(12, 6)); hv.line_hpi_all(df); plt.show()
```

## 2. Largest HPI change between first and last year

```python
plt.figure(figsize=(9, 5)); hv.top_growth(df, top_n=5); plt.show()
```

## 3. Affordability ratio: ten least affordable markets each year

```python
g = sns.FacetGrid(df, col="Year", col_wrap=3, height=4, sharex=False)
g.map_dataframe(hv.affordability_bar, n_countries=10)
plt.tight_layout(); plt.show()
```

## 4. Smoothed HPI (3-year centred rolling mean)

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 7), sharey=True)
for ax, c in zip(axes.ravel(), ["USA", "Japan", "Germany", "Australia"]):
    hv.plot_hpi_smoothed(df, c, ax=ax)
plt.tight_layout(); plt.show()
```

## 5. Year-over-year % change in HPI

```python
plt.figure(figsize=(10, 6)); hv.yoy_heatmap(df); plt.show()
```

## 6. Affordability by region

Several regions hold a single country (Africa, Europe/Asia, Middle East, Oceania), so those boxes are one country's ten years, not a regional picture.

```python
plt.figure(figsize=(9, 5)); hv.region_boxplot(df); plt.show()
```

## 7. Correlation matrix for one country

Each coefficient is based on only 10 yearly points, so values around ±0.5 are well within what noise produces.

```python
vars_num = ["House_Price_Index", "Rent_Index", "Mortgage_Rate_pct",
            "Inflation_Rate_pct", "GDP_Growth_pct", "Population_Growth_pct"]
plt.figure(figsize=(6, 5)); hv.corr_heatmap(df, "Japan", vars_num); plt.show()
```

## 8. Interactive affordability slider (Plotly)

```python
hv.plotly_affordability_slider(df)
```

## 9. Does the data behave like real housing data?

Real indicators are persistent (next year looks like this year) and countries differ. Random draws are not, and do not.
